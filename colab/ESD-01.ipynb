```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq


```

```python

# Parámetros
fs = 1000        # Hz
T = 1.0          # s
N = int(fs * T)
t = np.arange(N) / fs

# Señal en voltios
f0 = 50          # Hz
signal = np.sin(2 * np.pi * f0 * t) + 0.2 * np.random.randn(N)

# FFT
X = fft(signal)
freqs = fftfreq(N, 1/fs)

# Densidad espectral de energía
ESD = np.abs(X)**2 / N

# Solo frecuencias positivas
idx = freqs >= 0
freqs = freqs[idx]
ESD = ESD[idx]


```

```python
# Gráficas
plt.figure(figsize=(10, 6))

# Tiempo
plt.subplot(2, 1, 1)
plt.plot(t, signal)
plt.xlabel("Tiempo [s]")
plt.ylabel("Voltaje [V]")
plt.title("Señal en el dominio del tiempo")
plt.grid(True)

# Frecuencia
plt.subplot(2, 1, 2)
plt.plot(freqs, ESD)
plt.xlabel("Frecuencia [Hz]")
plt.ylabel("Densidad espectral de energía [V²/Hz]")
plt.title("Densidad espectral de energía (ESD)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Parámetros
fs = 1000        # Hz
T = 1.0          # s
N = int(fs * T)
t = np.arange(N) / fs

# Escalón truncado (1 V)
step = np.zeros(N)
step[int(N/4):] = 1.0   # escalón en t = 0.25 s

# FFT
X = fft(step)
freqs = fftfreq(N, 1/fs)

# Densidad espectral de energía
ESD = np.abs(X)**2 / N

# Frecuencias positivas
idx = freqs >= 0
freqs = freqs[idx]
ESD = ESD[idx]
```

```python
# Gráficas
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.plot(t, step)
plt.xlabel("Tiempo [s]")
plt.ylabel("Voltaje [V]")
plt.title("Escalón truncado en el tiempo")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(freqs, ESD)
plt.xlabel("Frecuencia [Hz]")
plt.ylabel("Densidad espectral de energía [V²/Hz]")
plt.title("ESD del escalón truncado")
plt.grid(True)

plt.tight_layout()
plt.show()
```
