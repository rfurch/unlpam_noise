```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parámetros de la señal y cuantización
frecuencia = 1        # Hz
amplitud = 1
muestras = 200         # Cuántos puntos en total
niveles = 64           # Niveles de cuantización (ej: 3 bits -> 2^3 = 8)

# 2. Generar el eje x y la señal seno original
t = np.linspace(0, 2, muestras)
seno_original = amplitud * np.sin(2 * np.pi * frecuencia * t)

# 3. Cuantizar la señal
# Se escala, redondea y se reescala para asignar valores discretos
seno_cuantizado = np.round(seno_original * (niveles / 2)) / (niveles / 2)

# 4. Graficar
plt.figure(figsize=(10, 6))
plt.plot(t, seno_original, label='Original', alpha=0.5, linestyle='--')
plt.step(t, seno_cuantizado, label='Cuantizada', where='mid', color='red', linewidth=2)
plt.title('Función Seno Cuantizada')
plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.grid(True)
plt.legend()
plt.show()
```
