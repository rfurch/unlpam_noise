```python
# Importaciones necesarias (librerias)

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Parámetros ajustables y armado de la senal

fs = 1000          # Frecuencia de muestreo (Hz)
T = 1.0            # Duración (s)
f0 = 50            # Frecuencia de la señal (Hz)

t = np.arange(0, T, 1/fs)

# Señal en el tiempo
signal = np.sin(2 * np.pi * f0 * t)
signal = np.sin(2*np.pi*50*t) + 0.5*np.sin(2*np.pi*145*t)

# Ruido
noise = 0.1 * np.random.randn(len(t))

# Señal total
x = signal # + noise

# FFT
X = np.fft.fft(x)
freqs = np.fft.fftfreq(len(X), 1/fs)

# Solo parte positiva
idx = freqs >= 0
freqs = freqs[idx]
X = X[idx]

# Magnitud normalizada
X_mag = np.abs(X) / len(x)

```

```python
# --- Gráficas iniciales
plt.figure(figsize=(12, 6))

# Dominio del tiempo
plt.subplot(2, 1, 1)
plt.plot(t, x)
plt.title("Señal en el dominio del tiempo")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.grid(True)

# Dominio de la frecuencia
plt.subplot(2, 1, 2)
plt.stem(freqs, X_mag, basefmt=" ")
plt.title("Espectro de frecuencia (FFT)")
plt.xlabel("Frecuencia [Hz]")
plt.ylabel("Magnitud")
plt.xlim(0, 200)
plt.grid(True)

plt.tight_layout()
plt.show()
```
