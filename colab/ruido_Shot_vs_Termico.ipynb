```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Parámetros
k = 1.380649e-23       # Constante de Boltzmann
q = 1.602e-19          # Carga del electrón

T = 300                # Kelvin
R = 10e2               # 10 kΩ
I = 10e-4              # 10 µA

fs = 10_000            # Hz
N = 5000
t = np.arange(N) / fs
```

```python
# Ruido térmico (voltaje)
sigma_thermal = np.sqrt(4 * k * T * R * fs / 2)
thermal_noise = sigma_thermal * np.random.randn(N)

# Ruido shot (corriente → voltaje)
sigma_shot_i = np.sqrt(2 * q * I * fs / 2)
shot_noise_i = sigma_shot_i * np.random.randn(N)
shot_noise_v = shot_noise_i * R   # Convertimos a voltaje
```

```python
# Gráficas
plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(t, thermal_noise * 1e6)
plt.title("Ruido térmico (Johnson)")
plt.ylabel("Voltaje [µV]")
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(t, shot_noise_v * 1e6)
plt.title("Ruido Shot")
plt.xlabel("Tiempo [s]")
plt.ylabel("Voltaje [µV]")
plt.grid()

plt.tight_layout()
plt.show()
```
