```python
## ejemplos de ventana

# Importaciones necesarias (librerias)

import numpy as np
import matplotlib.pyplot as plt
```

```python
## ejemplos de ventana

fs = 1000
T = 1.0
f0 = 53.5  # frecuencia NO entera → provoca leakage

t = np.arange(0, T, 1/fs)
x = np.sin(2 * np.pi * f0 * t)

# Ventanas
rect = np.ones(len(x))
hann = np.hanning(len(x))

# Señales
x_rect = x * rect
x_hann = x * hann

def fft_mag(x):
    X = np.fft.fft(x)
    freqs = np.fft.fftfreq(len(x), 1/fs)
    idx = freqs >= 0
    return freqs[idx], np.abs(X[idx]) / len(x)

f_rect, X_rect = fft_mag(x_rect)
f_hann, X_hann = fft_mag(x_hann)

```

```python
# --- Gráficas ---
plt.figure(figsize=(12, 6))


# Dominio del tiempo
plt.subplot(3, 1, 1)
plt.plot(t, x)
plt.title("Señal en el dominio del tiempo")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(f_rect, X_rect)
plt.title("FFT sin ventana (rectangular)")
plt.xlim(40, 60)
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(f_hann, X_hann)
plt.title("FFT con ventana Hann")
plt.xlim(40, 60)
plt.grid(True)

plt.tight_layout()
plt.show()
```
